```python
# Import library
import matplotlib.pyplot as plt

# Data
categories = ['Yes', 'No']
values = [28, 35]
#edit these values to match your survey data

# Plot
plt.bar(categories, values)
plt.xlabel('Can you play a musical instrument?') #edit this xlabel to match your yes/no question, make sure to leave the quote symbols.
plt.ylabel('Frequency')
plt.show()
```
